```python
from py3plex.core import multinet
import itertools
import matplotlib.pyplot as plt
from py3plex.visualization.multilayer import hairball_plot
```

```python
network = multinet.multi_layer_network(network_type="multiplex")
```

```python
nodos = range(1, 9)
```

```python
edges_trabajo = [
    {"source": 1, "source_type": "trabajo", "target": 2, "target_type": "trabajo", "weight": 1},
    {"source": 1, "source_type": "trabajo", "target": 3, "target_type": "trabajo", "weight": 1},
    {"source": 1, "source_type": "trabajo", "target": 4, "target_type": "trabajo", "weight": 1},
    {"source": 2, "source_type": "trabajo", "target": 5, "target_type": "trabajo", "weight": 1},
    {"source": 2, "source_type": "trabajo", "target": 6, "target_type": "trabajo", "weight": 1},
    {"source": 4, "source_type": "trabajo", "target": 7, "target_type": "trabajo", "weight": 1},
    {"source": 4, "source_type": "trabajo", "target": 8, "target_type": "trabajo", "weight": 1}
]
```

```python
network.add_edges(edges_trabajo)
```

```python
edges_vivienda = [
    {"source": u, "source_type": "Vivienda", "target": v, "target_type": "Vivienda", "weight": 1}
    for u, v in itertools.combinations(nodos, 2)
]
network.add_edges(edges_vivienda)
```

```python
edges_intercapa = [
    {"source": i, "source_type": "trabajo", "target": i, "target_type": "Vivienda", "weight": 1}
    for i in nodos
]
network.add_edges(edges_intercapa)
```

```python
plt.figure(figsize=(10, 8))

# Dibujamos la red
hairball_plot(
    network.core_network,
    color_list=["#1f77b4", "#ff7f0e", "purple"],
    layout_parameters={"iterations": 100}
)

plt.title("Red Multiplex: Trabajo y Vivienda")
plt.show()
```
